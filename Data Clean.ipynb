```python
import matplotlib.pyplot as plt
import numpy as np
fig = plt.figure()
ax = plt.axes()
```

```python
top_Products = df.groupby(['product_name','country'])['sales'].sum().reset_index().nlargest(10,'sales')
top_Products = top_Products.sort_values(by='sales', ascending=True)
colors = plt.cm.Blues(top_Products['sales']/top_Products['sales'].max())
plt.figure(figsize=(10,5))
plt.barh(top_Products['product_name'],top_Products['sales'], color = colors)
plt.xlabel('sales',color="green",size=12)
plt.ylabel('Products Name',color="green",size=12)
plt.title('Top 10 Proudcts by sales',color="green",size=15)
plt.show()
```

```python
top_profit = df.groupby(['state'])['profit'].sum().reset_index().nlargest(10,'profit')
top_profit = top_profit.sort_values(by='profit', ascending=True)
colors = plt.cm.Reds(top_profit['profit'] / top_profit['profit'].max())
plt.figure(figsize=(10,5))
plt.barh(top_profit['state'],top_profit['profit'], color=colors)
plt.xlabel('profit',color="red",size=12 )
plt.ylabel('state',color="red",size=12)
plt.title('Top 10 state by profit',color="red",size=15)
plt.show()
```

```python
top_customerName = df.groupby(['customer_name'])['sales'].sum().reset_index().nlargest(10,'sales')
top_customerName = top_customerName.sort_values(by='sales', ascending=True)
colors = plt.cm.Greens(top_customerName['sales'] / top_customerName['sales'].max())
plt.figure(figsize=(10,4))
plt.barh(top_customerName['customer_name'],top_customerName['sales'], color = colors)
plt.xlabel('sales',color="green",size=12)
plt.ylabel('Products Name',color="green",size=12)
plt.title('Top 10 customer by sales',color="green",size=15)
plt.show()
```

```python
bottom_category = df.groupby(['category'])['sales'].mean().reset_index().nsmallest(3,'sales')
bottom_category = bottom_category.sort_values(by='sales', ascending=False)
colors = plt.cm.YlOrRd(bottom_category['sales'] / bottom_category['sales'].max())
plt.figure(figsize=(10,4))
plt.barh(bottom_category['category'],bottom_category['sales'], color = colors)
plt.xlabel('sales',color="blue",size=12)
plt.ylabel('Sub-Categories Name',color="blue",size=12)
plt.title('Bottom 3 Sub-Categories by Average Sales',color="blue",size=15)
plt.show()
```
